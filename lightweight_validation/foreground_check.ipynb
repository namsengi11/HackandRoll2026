```python
# Convert all HEIC assets to jpg

from pathlib import Path
from PIL import Image
import pillow_heif

pillow_heif.register_heif_opener()

assets_dir = Path("assets/foreground-check")
for path in assets_dir.iterdir():
  if path.suffix.lower() == ".heic":
    image = Image.open(path)
    jpg_path = path.with_suffix(".jpg")
    image.save(jpg_path, format="JPEG")
    path.unlink()
```

```python
image_path = "assets/foreground-check/IMG_6422 2.jpg"
image_bytes = open(image_path, "rb").read()
check_foreground(image_bytes)
```

```python
segmentation = render_segmentation(image_bytes)

from IPython.display import Image, display
display(Image(segmentation))
```
